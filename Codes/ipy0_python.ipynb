# Lists and dictionaries

```python
# A list is an ordered sequence of elements:
mylist = ['cat', 'window', 3]  # remember numbering starts from 0
print(mylist[0]) # first element of the list

# A dictionary stores key–value pairs:
person = {'name': 'John', 'age': 35, 'city': 'Milan'}
print(person['name'])    # 'John'
print(person['age'])     # 35
print(person['city'])    # 'Milan'
```

# Functions

```python
# Definition
def fib(n):    # write Fibonacci series less than n
    a, b = 0, 1
    while a < n:
        print(a, end=' ')
        a, b = b, a+b
    print()

# Now call the function we just defined:
fib(2000)
```

# Classes

A class in Python is a blueprint (or template) for creating objects. It defines attributes (data) and methods (functions) that the objects created from it will have. For instance:


```python
class Point:

    def __init__(self, x, y):   # initialisation
        self.x = x
        self.y = y

    def move(self, dx, dy):   # method
        self.x += dx
        self.y += dy

p=Point(1,2)
p.move(3, -1)
print(p.x, p.y)         
```

As we will very often define classes, we use *@dataclass* which is a *class decorator*, which is easier to use and interpret. This has many applications in *readgmsh*

```python
from dataclasses import dataclass

@dataclass
class Point:
    x: int
    y: int

    def move(self, dx: int, dy: int):
        self.x += dx
        self.y += dy

p = Point(1, 2)
p.move(3, -1)
print(p.x, p.y)    # 4 1
```

# NumPy

NumPy is the standard library for numerical computing in Python. Its main advantages are speed, convenience, and powerful mathematical operations.
Here are the main reasons to use it.

1. Much faster than Python lists. NumPy arrays are stored in contiguous memory and implemented in optimized C code.
2. Vectorized operations. You can perform operations on an entire array without writing loops.
3. Multi-dimensional arrays. NumPy naturally handles matrices and higher-dimensional arrays.



```python
import numpy as np

a = np.array([1, 2, 3, 4]) # defines first array
print(a[0:-2],': all entries from beginning to end excluding last two\n')  
print(a ** 2, ': squares all entries at once\n')  

b = np.array([5, 6, 7, 8]) # defines second array 
print(a @ b,': dot product\n') 
print(a * b, ': element-by-element multiplication\n') 

c = np.array([[1, 2, 3],
              [4, 5, 6]])  # (2x3) matrix
print(c[0][1],': generic element\n') 
print(c.shape,': dimensions of the matrix c\n')      

d = np.array([[2, 3],
              [5, 6]])  # (2x2) matrix

print(c.T @ d, 'matrix multiplcation between c tranposed and d\n')
```
